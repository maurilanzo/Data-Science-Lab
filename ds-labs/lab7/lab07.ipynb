```python
import numpy as np
import pandas as pd
```

```python
!wget https://github.com/dbdmg/data-science-lab/raw/master/datasets/online_retail.csv
```

```python
dataset = pd.read_csv("online_retail.csv")
```

```python
dataset = dataset[~(np.strings.startswith(dataset["InvoiceNo"].values.astype(str),"C"))]
```

```python
pivot = pd.pivot_table(dataset,index = "InvoiceNo",values= "Description",aggfunc=np.unique)
```
