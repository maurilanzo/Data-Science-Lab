```python
#padding before fft
import os
import numpy as np
import pandas as pd
from scipy.io import wavfile

# === ORIGINAL STRUCTURE PRESERVED ===
wavs_dev = ((os.listdir("free-spoken-digit/dev")))
wavlist_dev = sorted(wavs_dev, key=lambda x: int(x.split("_")[0]))
devs = []
lengs = []
X_dev = []
y_dev = []

TARGET_LENGTH = 4000  # Number of samples to trim/pad BEFORE FFT

for file in wavlist_dev:
    path = os.path.join("free-spoken-digit/dev", file)

    parts = file.split('_')
    label = int(parts[1].split('.')[0])
    print(label)
    y_dev.append(label)
    
    rate, signal = wavfile.read(path)

    # === TRIM OR PAD SIGNAL BEFORE FFT ===
    if signal.ndim > 1:
        signal = signal[:, 0]  # Convert to mono if stereo

    if len(signal) > TARGET_LENGTH:
        signal = signal[:TARGET_LENGTH]
    else:
        pad_width = TARGET_LENGTH - len(signal)
        signal = np.pad(signal, (0, pad_width), mode='constant')

    # === NOW DO FFT ON FIXED-LENGTH SIGNAL ===
    fft_val = np.abs(np.fft.fft(signal))

    X_dev.append(fft_val)

X_train = pd.DataFrame(X_dev)
y_train = pd.DataFrame(y_dev)
```
